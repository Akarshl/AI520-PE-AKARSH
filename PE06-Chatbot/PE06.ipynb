```python
# Setup and Imports
import nltk
from nltk.tokenize import word_tokenize
import random
from textblob import TextBlob
```

```python
# Download required NLTK data (run once)
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# Core Dataset
dataset = {
    "hello": "Hi there! How can I help you today?",
    "how are you": "I'm a bot, so I don't have feelings, but thanks for asking!",
    "what is your name": "I'm a simple chatbot created using NLTK.",
    "tell me a joke": "Why did the computer show up at work late? It had a hard drive!",
    "what is the capital of france": "The capital of France is Paris."
}
```

```python
# Extended Dataset - Adding 5 new question-answer pairs
extended_questions = {
    "who created you": "I was created by a Python developer like you!",
    "what can you do": "I can chat with you and answer simple questions.",
    "are you intelligent": "I try my best, but I'm just a rule-based bot.",
    "do you have feelings": "I don't have feelings, but I understand yours!",
    "tell me something funny": "Why don’t programmers like nature? It has too many bugs!"
}
dataset.update(extended_questions)
```

```python
# Core Functions

def handle_user_input(user_input):
    """
    Exact match or close match for user input in dataset.
    """
    user_tokens = set(word_tokenize(user_input.lower()))
    # Try exact match first (lowercased)
    if user_input.lower() in dataset:
        return dataset[user_input.lower()]
    
    # If no exact match, try partial matching (intersection)
    best_match = None
    max_overlap = 0
    for question in dataset:
        question_tokens = set(word_tokenize(question))
        overlap = len(user_tokens.intersection(question_tokens))
        if overlap > max_overlap:
            max_overlap = overlap
            best_match = question

    # Heuristic: If overlap is significant (>0), return the best matched answer
    if max_overlap > 0:
        return dataset[best_match]

    return None

def handle_ambiguous_input(user_input):
    """
    Handle inputs with no exact match using partial matching.
    """
    user_tokens = set(word_tokenize(user_input.lower()))
    partial_matches = []
    for question in dataset:
        question_tokens = set(word_tokenize(question))
        if user_tokens.intersection(question_tokens):
            partial_matches.append(dataset[question])
    if partial_matches:
        return random.choice(partial_matches)
    else:
        return "I'm sorry, I didn't understand that. Can you rephrase?"

def analyze_sentiment(text):
    """
    Use TextBlob to analyze sentiment.
    Returns polarity (-1 to 1) and subjectivity (0 to 1)
    """
    blob = TextBlob(text)
    sentiment = blob.sentiment
    return sentiment.polarity, sentiment.subjectivity

def enhanced_handle_user_input(user_input):
    """
    Combine regular response with sentiment analysis.
    """
    response = handle_user_input(user_input)
    if response is None:
        response = handle_ambiguous_input(user_input)
    polarity, subjectivity = analyze_sentiment(user_input)

    sentiment_summary = ""
    if polarity > 0.1:
        sentiment_summary = "Your message seems positive."
    elif polarity < -0.1:
        sentiment_summary = "Your message seems negative."
    else:
        sentiment_summary = "Your message seems neutral."

    return f"{response} ({sentiment_summary})"

def start_chat():
    print("Welcome to the NLTK rule-based chatbot! (type 'exit' or 'quit' to stop)")
    while True:
        user_input = input("You: ").strip()
        if user_input.lower() in ['exit', 'quit', 'bye', 'stop']:
            print("Chatbot: Goodbye! Have a nice day!")
            break
        response = enhanced_handle_user_input(user_input)
        print("Chatbot:", response)
```

```python
# Testing extended questions
def test_extended_questions():
    for question in extended_questions:
        response = enhanced_handle_user_input(question)
        print(f"Q: {question}\nA: {response}\n")

# To run the chatbot:
if __name__ == "__main__":
    # Uncomment below to test extended dataset
    # test_extended_questions()
    
    start_chat()
```
